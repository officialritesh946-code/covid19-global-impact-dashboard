# COVID-19 Global Impact Dashboard
Historical global COVID-19 trend and country/region analysis.

```python
import os, numpy as np, pandas as pd
import matplotlib.pyplot as plt
RAW='raw_data'
daily=pd.read_csv(os.path.join(RAW,'day_wise.csv'), parse_dates=['Date'])
country=pd.read_csv(os.path.join(RAW,'country_wise_latest.csv'))
daily.head()
```

## 1. Derived metrics

```python
daily['7D Avg New Cases']=daily['New cases'].rolling(7).mean()
daily['7D Avg New Deaths']=daily['New deaths'].rolling(7).mean()
daily['Case Fatality Rate (%)']=daily['Deaths']/daily['Confirmed']*100
prev7=daily['Confirmed'].shift(7)
ratio=daily['Confirmed']/prev7
daily['Doubling Time (days)']=np.where((prev7>0)&(ratio>1),7*np.log(2)/np.log(ratio),np.nan)
daily.tail()
```

## 2. Global trend

```python
plt.figure(figsize=(12,5))
plt.plot(daily['Date'],daily['7D Avg New Cases'])
plt.title('Global new cases — 7-day rolling average')
plt.xlabel('Date'); plt.ylabel('Cases'); plt.grid(alpha=.2); plt.show()
```

## 3. Country analysis

```python
country['Case Fatality Rate (%)']=country['Deaths']/country['Confirmed']*100
country['Recovery Rate (%)']=country['Recovered']/country['Confirmed']*100
country.nlargest(10,'Confirmed')[['Country/Region','Confirmed','Deaths','Recovered','Active']]
```

## 4. Region summary

```python
region=country.groupby('WHO Region',as_index=False)[['Confirmed','Deaths','Recovered','Active']].sum()
region.sort_values('Confirmed',ascending=False)
```

## 5. Interpretation
The supplied dataset is historical through 27 July 2020. It does not include vaccination variables, so vaccination rollout is not inferred or fabricated.
